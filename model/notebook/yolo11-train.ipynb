```python
# ============================================================
# CREATE LOCAL DATA.YAML FOR THE FRUITS DETECTION DATASET
# ============================================================

from pathlib import Path
import yaml

FRUIT_DATASET = Path(
    r"C:\major-project\model\yolo11\Fruits-detection"
)

data_config = {
    "path": str(FRUIT_DATASET),
    "train": "train/images",
    "val": "valid/images",
    "test": "test/images",
    "names": {
        0: "Apple",
        1: "Banana",
        2: "Grape",
        3: "Orange",
        4: "Pineapple",
        5: "Watermelon"
    }
}

LOCAL_DATA_YAML = FRUIT_DATASET / "data_local.yaml"

with open(LOCAL_DATA_YAML, "w") as f:
    yaml.dump(data_config, f, sort_keys=False)

print("Local data.yaml created successfully!")
print("Path:", LOCAL_DATA_YAML)

print("\nConfiguration:")
print(data_config)
```

```python
# ============================================================
# VERIFY FRUITS DETECTION DATASET
# ============================================================

from pathlib import Path

for split in ["train", "valid", "test"]:
    images_dir = FRUIT_DATASET / split / "images"
    labels_dir = FRUIT_DATASET / split / "labels"

    image_count = len(list(images_dir.glob("*.*"))) if images_dir.exists() else 0
    label_count = len(list(labels_dir.glob("*.txt"))) if labels_dir.exists() else 0

    print(f"\n{split.upper()}")
    print("Images folder exists:", images_dir.exists())
    print("Labels folder exists:", labels_dir.exists())
    print("Images:", image_count)
    print("Labels:", label_count)
```

```python
# ============================================================
# TRAIN APPLE-ONLY DETECTION MODEL USING GPU
# ============================================================

from ultralytics import YOLO
from pathlib import Path

# Apple-only dataset
DATA_YAML = Path(r"C:\major-project\model\yolo11_dataset\data.yaml")

# Use the pretrained model that already exists on your computer
MODEL_PATH = Path(
    r"C:\major-project\model\yolo11\Fruits-detection\yolov8s.pt"
)

# Check files before training
assert DATA_YAML.exists(), f"Dataset YAML not found: {DATA_YAML}"
assert MODEL_PATH.exists(), f"Model not found: {MODEL_PATH}"

# Load local pretrained model
model = YOLO(str(MODEL_PATH))

# Train ONLY on the Apple dataset
results = model.train(
    data=str(DATA_YAML),
    epochs=100,
    imgsz=640,
    batch=8,
    patience=20,
    device=0,
    project=r"C:\major-project\model\yolo11\runs",
    name="apple_detection_gpu",
    exist_ok=True,
    plots=True
)

print("\nApple-only model training completed!")
```

```python
# ============================================================
# EVALUATE APPLE-ONLY MODEL ON TEST DATA
# ============================================================

from ultralytics import YOLO

# Paths
BEST_MODEL_PATH = r"C:\major-project\model\yolo11\runs\apple_detection_gpu\weights\best.pt"
DATA_YAML = r"C:\major-project\model\yolo11_dataset\data.yaml"

# Load the best Apple-only model
best_model = YOLO(BEST_MODEL_PATH)

# Evaluate on the TEST dataset
metrics = best_model.val(
    data=DATA_YAML,
    split="test",
    imgsz=640,
    device=0,
    plots=True
)

# Print accuracy metrics
print("\n========== APPLE DETECTION TEST RESULTS ==========")
print(f"Precision:  {metrics.box.mp * 100:.2f}%")
print(f"Recall:     {metrics.box.mr * 100:.2f}%")
print(f"mAP@50:     {metrics.box.map50 * 100:.2f}%")
print(f"mAP@50-95:  {metrics.box.map * 100:.2f}%")
```
